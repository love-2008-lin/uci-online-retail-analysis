# 每日交易汇总核验

[每日汇总 SQL](../sql/01_daily_and_market_summary.sql)将 57,834 行明细聚合成 26 个有交易日期，再与 10 月日历左连接。输出覆盖全部 31 天，10 月 1、8、15、22、29 日的金额和发票数均补零。

用 Pandas 抽查 10 月 2 日，有效明细金额为 £11,428.59，去重发票数为 35，与 SQL 结果一致。

```python
import pandas as pd
df = pd.read_csv('../data/processed/data_10month.csv')
```

```python
df['invoicedate'] = pd.to_datetime(df['invoicedate'])
df_grouped_money = df.groupby(df['invoicedate'].dt.normalize())['lineamount'].sum()
df_grouped_count = df.groupby(df['invoicedate'].dt.normalize())['invoiceno'].nunique()
print(df_grouped_money.head(2))
print(df_grouped_count.head(2))
```
