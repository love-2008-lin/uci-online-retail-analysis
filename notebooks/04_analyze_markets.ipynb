# 市场交易分析

统计正向商品金额、有效发票数和平均每张发票金额。

```python
import pandas as pd
data = pd.read_csv('../data/processed/data_10month.csv')
data.head(5)
```

```python
def data_country(data: pd.DataFrame) -> pd.DataFrame:
    """返回按正向商品金额降序排列的国家交易汇总。"""
    data_cp = data.copy()
    data_cp['country'] = data_cp['country'].fillna('uncertain')
    data_countrys = data_cp.groupby("country").agg(
        gross_amount_gbp = ('lineamount','sum'),
        invoice_count = ('invoiceno','nunique')
    ).reset_index().sort_values('gross_amount_gbp',ascending=False)
    data_countrys['avg_invoice_gbp'] = data_countrys['gross_amount_gbp'] / data_countrys['invoice_count']
    return data_countrys


data_countrym = data_country(data)
display(data_countrym.head(5))
```
